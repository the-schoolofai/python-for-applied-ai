## Part B · Lab 1 — your first notebook

### Cell 1

````python
name = "Bilal"
city = "Lahore"

print("Hello,", name + "!", "Welcome from", city + "!")
````

**What it prints:**

```
Hello, Bilal! Welcome from Lahore!
```

Two things are happening in that one line, and they are not the same thing:

- `name + "!"` **joins** two pieces of text together, with nothing in between: `Bilal!`
- The commas inside `print(...)` **separate** the things being printed, and print puts one space between them.

That is why there is no space before the exclamation mark, but there is one before `Welcome`.


### Cell 2

````python
orders = [120, 340, 90, 560]

print("Order Summary")
print("--------------")
print("Total Orders:", len(orders))
print("Total Amount:", sum(orders))
print("Biggest Order:", max(orders))
````

**What it prints:**

```
Order Summary
--------------
Total Orders: 4
Total Amount: 1110
Biggest Order: 560
```

`len` counts **how many** items are in the list — four orders. It does not add them up.
`sum` adds the values. `max` finds the largest one.

## Part C · Break it on purpose

**The three cells below are meant to fail.** Run each one, read the error, then look at the fix underneath.

Remember: the **last line** of an error names the problem. Everything above it is the path Python took to get there.

### Error 1 · Deleted a quotation mark

````python
# MEANT TO FAIL — the closing quote after Lahore is missing
city = "Lahore

print(city)
````

**The last line says something like:**

```
SyntaxError: unterminated string literal (detected at line 2)
```

**In plain words:** a piece of text was opened with a quote mark and never closed, so Python reached the end of the line still waiting for it.

**The fix:** put the closing quote back.

````python
city = "Lahore"

print(city)
````

### Error 2 · Misspelled a variable name

````python
# MEANT TO FAIL — the variable is called city, not citty
city = "Lahore"

print(citty)
````

**The last line says:**

```
NameError: name 'citty' is not defined
```

**In plain words:** Python looked for a box with that label and there is no such box. Nothing was ever put in `citty`.

This is also the error you get from a **capital letter in the wrong place** — `City` is a different name from `city`. Python is case-sensitive, always.

**The fix:** spell it exactly as you created it.

````python
city = "Lahore"

print(city)
````

### Error 3 · Text plus a number

````python
# MEANT TO FAIL — you cannot join text to a number with +
print("Total Orders:" + 120)
````

**The last line says:**

```
TypeError: can only concatenate str (not "int") to str
```

**In plain words:** the `+` sign means *join these two pieces of text*. One side is text and the other is a number, so Python does not know what you want.

There are two correct fixes, and they are both worth knowing.

````python
# Fix 1 — use a comma. print adds a space between the parts.
print("Total Orders:", 120)

# Fix 2 — turn the number into text first, then join it.
print("Total Orders: " + str(120))
````

Both print exactly the same line. The difference is in the code, not the result: fix 1 gets the space from `print` for free, while fix 2 only has a space because one was typed inside the quotes.

**Which error was hardest?** For most people it is number 3, because the words *concatenate*, *str* and *int* are unfamiliar on day one. After today they are not.

## Part D · Lab 2 — something actually useful

````python
names = ["  ali khan ", "SARA AHMED", "bilal  ", " Ayesha Khan"]

for raw in names:
    clean = raw.strip().title()
    print(clean)
````

**What it prints:**

```
Ali Khan
Sara Ahmed
Bilal
Ayesha Khan
```

Four messy names went in. Four clean ones came out, and you did not touch them one at a time.


### Now change things and see what happens

Each variation is in its own cell below, with the result explained. **Run them** — do not just read them.

#### Replace `.title()` with `.upper()`

````python
names = ["  ali khan ", "SARA AHMED", "bilal  ", " Ayesha Khan"]

for raw in names:
    print(raw.strip().upper())
````

Everything comes out in capitals: `ALI KHAN`, `SARA AHMED`, `BILAL`, `AYESHA KHAN`.

`.title()` capitalises the first letter of each word. `.upper()` capitalises every letter.

#### Remove `.strip()`

````python
names = ["  ali khan ", "SARA AHMED", "bilal  ", " Ayesha Khan"]

for raw in names:
    print(raw.title())
````

Look closely at the output. The spaces are back — `  Ali Khan` is pushed to the right, and `Bilal  ` has two invisible spaces after it that you cannot see but a computer can.

This matters more than it looks. `"Bilal"` and `"Bilal  "` are **not equal** as far as Python is concerned, so a name with a stray space will not match, will not join to another table, and will quietly become a separate customer in a spreadsheet.

#### Add your own messy name to the list

````python
names = ["  ali khan ", "SARA AHMED", "bilal  ", " Ayesha Khan", "   fatima  ZAHRA  "]

for raw in names:
    print(raw.strip().title())
````

It handles the new one exactly the same way: `Fatima  Zahra`.

But notice the **double space still in the middle** — `.strip()` only removes spaces at the two ends, never inside. Fixing that needs `.replace("  ", " ")`, which is Day 02.

#### Remove the indentation on the last line

````python
# MEANT TO FAIL — print is no longer inside the loop
names = ["  ali khan ", "SARA AHMED", "bilal  "]

for raw in names:
clean = raw.strip().title()
print(clean)
````

```
IndentationError: expected an indented block after 'for' statement on line 4
```

**In plain words:** the `for` line ends with a colon, which promises that something indented comes next. Nothing was indented, so Python stopped.

Indentation is not decoration in Python. It is how the language knows which lines belong inside the loop.


### The two written answers

**What does `.strip()` do?**
> It removes any spaces at the beginning and the end of a piece of text. It does not touch spaces in the middle.

**What does the `for` line do?**
> It takes the items of the list one at a time, calls each one `raw` while it works on it, and runs the indented lines underneath once for every item.

## Part E · When it will not run — the five checks

| # | Check | What it looks like |
|---|---|---|
| 1 | A capital letter | `Print` instead of `print`, or `Orders` instead of `orders` |
| 2 | A missing quotation mark | The colours in the cell look wrong from that point on |
| 3 | A missing colon | Nothing after `for raw in names` |
| 4 | Indentation | The lines under `for` are not pushed in, or not pushed in equally |
| 5 | Cells run out of order | You ran Cell 2 before Cell 1. Run from the top |

Number 5 is the one that confuses people most in a notebook, because the code **looks** right.

The cell below empties the box on purpose first, which is exactly what **Runtime → Restart** does to every variable you have made.

````python
# del empties the box — the same effect a restart has on every variable
del city

print(city)
````

```

```

Nothing is wrong with `print(city)`. The box is simply empty, because the cell that filled it has not been run in this session.
